```python
#Q1
import pandas as pd
a = np.array([82,84,81,83,85])
b = np.array([85,87,86,84,88])
c = np.array([88,89,91,90,92])
a_mean = np.mean(a)
b_mean = np.mean(b)
c_mean = np.mean(c)
grand_mean = (a_mean+b_mean+c_mean)/3
SSB = ((a_mean-grand_mean)**2)*5+((b_mean-grand_mean)**2)*5+((c_mean-grand_mean)**2)*5
mean_SSB = SSB/2
print(mean_SSB)

SSW = np.sum((a-a_mean)**2)+np.sum((b-b_mean)**2)+np.sum((c-c_mean)**2)
mean_SSW = SSW/12
print(mean_SSW)
F = mean_SSB/mean_SSW
print(F)
```

```python
import numpy as np

matrix = np.array([[18, 7],
                   [14, 11],
                   [20, 5]])

alpha = 5.991
m = 3
n = 2

tempMatrix = []
tempChi = []
row_totals = matrix.sum(axis=1)
col_totals = matrix.sum(axis=0)
grand_total = matrix.sum()

for i in range(m):
    temp1 = []
    for j in range(n):
        expected = (row_totals[i] * col_totals[j]) / grand_total
        temp1.append(expected)
        chi_val = ((matrix[i, j] - expected) ** 2) / expected
        tempChi.append(chi_val)

    tempMatrix.append(temp1)

print("The expected value is:")
print(np.array(tempMatrix))

calculated_chi = sum(tempChi)
print(f"\nThe calculated chi square value is: {calculated_chi:.4f}")

if calculated_chi < alpha:
    print("The null hypothesis is accepted")
else:
    print("The null hypothesis is rejected")
```
