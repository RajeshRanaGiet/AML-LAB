```python
import numpy as np

x1 = np.array([1,2,3,7,8,9])
x2 = np.array([2,3,4,6,7,8])
x3 = np.array([5,4,3,3,2,1])
y = np.array([0,0,0,1,1,1])

data = {'x1': x1, 'x2': x2, 'x3': x3, "y" : y}

def r(n,x,y):
  return abs((n*sum(x*y)-sum(x)*sum(y))/((n*sum(x**2)-(sum(x))**2) * (n*sum(y**2)-(sum(y))**2))**0.5)

def CFSMerit(k,x1,x2,y):
  rcf = (r(k,x1,y)+r(k,x2,y))/2
  rff = r(k,x1,x2)
  return (k*rcf)/(k+k*(k-1)*rff)

def bestMerit(data):
  temp = list(data.keys())
  comb = [(x,y) for idx,x in enumerate(temp[:-1]) for y in temp[idx:] if x!=y]
  n = len(data[temp[1]])

  temp = temp[:-1]
  resComb = [(x,y) for idx,x in enumerate(temp[:-1]) for y in temp[idx:] if x!=y]
  res = {}
  for i in resComb:
    res[i] = CFSMerit(n,data[i[0]],data[i[1]],data["y"])
  return max(res, key=res.get)

print("The best CFS Merit score is ", bestMerit(data))
```
